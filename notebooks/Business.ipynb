Question 5- Which areas are expensive?

```python

# avg_price_listing =df["price"].mean()
# max_price_listing = (
#     df[df["price"]>avg_price_listing]
#     # .sort_values(ascending=True)
# )

# print(max_price_listing.head(10))


# Calculate average price
avg_price_listing = df["price"].mean()

# Filter listings where price is greater than average price
max_price_listing = df[
    df["price"] > avg_price_listing
][["neighbourhood_cleansed", "price"]]

# Sort prices from highest to lowest
max_price_listing = max_price_listing.sort_values(
    by="price",
    ascending=False
)
# Display first 10 listings
print(max_price_listing.head(10))
```

Question 6- Are there unusual/extreme prices?

```python
# Select area and price
price_data = df[["neighbourhood_cleansed", "price"]]

# Lowest 10
print("Lowest 10 Prices:")
print(price_data.sort_values(by="price").head(10))

# Highest 10
print("\nHighest 10 Prices:")
print(price_data.sort_values(by="price", ascending=False).head(10))
```

Hosts Area  

Question 7- Which hosts manage the most listings?

```python
# Count the number of listings for each host
host_listings = (
    df.groupby(["host_id", "host_name"])
    .size()
    .sort_values(ascending=False) 
)
# Display the top 10 hosts with the most listings
print( host_listings.head(10))
```

 Question 8-How common are multi-listing hosts?

```python
# Count the number of hosts with only one listing
host_counts = (
    df.groupby(["host_id", "host_name"])
    .size()
    .sort_values(ascending=False)
)
# Print the count and percentage of hosts with only one listing
print("common multi-listing in count:",(host_counts == 1).sum())
print("common multi-listing in precentage:",(host_counts == 1).mean()*100)

```

Question 9- Where are they concentrated?

```python
# Find the neighbourhoods where multi-listing hosts are concentrated

multi_listing = df['neighbourhood_cleansed'].value_counts().head(10)

# Print the neighbourhoods with the most listings
print(multi_listing)
```

Performance Area 

Question 10- Which listings receive the most reviews?



```python
# Find the top-reviewed listings and sort them in descending order based on their number of reviews
top_reviewed = df[
    ["id",
     "host_name",
     "neighbourhood_cleansed",
     "number_of_reviews" ]
].sort_values(
    "number_of_reviews", 
    ascending=False )
# Display the top 10 reviewed listings
top_reviewed.head(10)
```

Question 11- Which neighbourhoods/room types have better ratings?

```python
# Find the top-rated listings and sort them in descending order based on their review scores
top_rating = df[
    ["id",
     "host_name",
     "neighbourhood_cleansed",
     "room_type",
     "review_scores_rating" ]
].sort_values(
    "review_scores_rating", 
    ascending=False )

# Display the top 10 rated listings
top_rating.head(10)
```


Question 12- Is price related to ratings or reviews?

```python
# Calculate average price, rating, and number of reviews
avg_price = df["price"].mean()
avg_rating = df["review_scores_rating"].mean()
avg_reviews = df["number_of_reviews"].mean()

# Print the averages of price, rating, and number of reviews
print("Average Price:", avg_price)
print("Average Rating:", avg_rating)
print("Average Reviews:", avg_reviews)
```

Availability Area 

Question 13- Which areas and room types have high or low availability?

```python
# Calculate the average occupancy across all listings
availability = (
    df.groupby(["neighbourhood_cleansed","room_type"])["estimated_occupancy_l365d"]
    .mean()
    .sort_values(ascending=False)
)

# Print the high and low availability listings
print("High availability:")
print(availability.head(10))  # High availability
print()
print("Low availability:")
print(availability.tail(10))   # Low availability
```

Question 14- What does availability suggest about the market?

```python
# Calculate the average occupancy across all listings
market_occupancy = df[
    "estimated_occupancy_l365d"
    ].mean()

# Print the average occupancy rounded to two decimal places
print("Average Occupancy:", round(market_occupancy, 2))
```

Opportunity Area 

Question 15- If a new host enters the NYC market,
which neighbourhood/room type and price range could be worth
considering based on the analysis?

```python
# Calculate the minimum, average, and maximum prices for each neighbourhood and room type for enters the market
result = (
    df.groupby(["neighbourhood_cleansed", "room_type"])
    ["price"]
      .agg(
          ["min", "mean", "max"])
      .reset_index()
)
# Print the results sorted by average price
print(result.sort_values("mean", ascending=True).head(10))
```
