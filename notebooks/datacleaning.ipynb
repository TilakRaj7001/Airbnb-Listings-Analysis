```python
import pandas as pd 
df = pd.read_csv(r"C:\Users\rahul\Desktop\AIRNB\data\rawdata\listings.csv")
```

```python
# Drop ALL unnecessary columns 
df.drop(columns=[
    "listing_url", "picture_url", "host_thumbnail_url", "host_picture_url",
    "host_profile_url", "host_url", "host_profile_id", "host_neighbourhood",
    "neighbourhood", "host_since", "calendar_last_scraped", "name",
    "description", "neighborhood_overview", "host_about", "host_location",
    "host_verifications", "host_total_listings_count", "host_has_profile_pic",
    "instant_bookable", "price_quote_checkin_date", "price_quote_checkout_date",
    "price_quote_total_price", "price_quote_price_per_night", "price_quote_raw",
    "calendar_updated", "has_availability", "availability_30", "availability_60",
    "availability_90", "availability_365", "availability_eoy", "minimum_nights",
    "maximum_nights", "minimum_minimum_nights", "maximum_minimum_nights",
    "minimum_maximum_nights", "maximum_maximum_nights", "minimum_nights_avg_ntm",
    "maximum_nights_avg_ntm", "calculated_host_listings_count", "license"
], inplace=True)
```

```python
# Drop column that 100% null values 
df=df.dropna(axis=1,how='all')
```

```python
import pandas as pd

# --- Calculate Missing Percentage ---
print("\n  MISSING VALUES PER COLUMN:")
missing_percent = df.isnull().mean() * 100
missing_cols = missing_percent[missing_percent > 0].sort_values(ascending=False)

if len(missing_cols) > 0:
    print(missing_cols)
    print(f"\nTotal columns with missing data: {len(missing_cols)}")
else:
    print("No missing values found in any column!")
```

```python
# checking datatype of all columns 
print(df.dtypes)
```

```python
# Fill missing values with "0" in int/float columns
# where the missing percentage is less than or equal to 5%
import pandas as pd
import numpy as np

for col in df.columns:

    missing_percentage = (df[col].isnull().sum() / len(df)) * 100

    if missing_percentage <= 5 and pd.api.types.is_numeric_dtype(df[col]):
        df[col] = df[col].fillna(0)
```

```python
# Fill missing values with "Unknown" in string/object columns
# where the missing percentage is less than or equal to 5%

for col in df.columns:

    missing_percentage = (df[col].isnull().sum() / len(df)) * 100

    if missing_percentage <= 5 and df[col].dtype == "string":
        df[col] = df[col].fillna("Unknown")
```

```python
import pandas as pd

# --- Calculate Missing Percentage ---
print("\n  MISSING VALUES PER COLUMN:")
missing_percent = df.isnull().mean() * 100
missing_cols = missing_percent[missing_percent > 0].sort_values(ascending=False)

if len(missing_cols) > 0:
    print(missing_cols)
    print(f"\nTotal columns with missing data: {len(missing_cols)}")
else:
    print("No missing values found in any column!")
```

```python
# checking datatype of all columns 
print(df['review_scores_location'].dtypes)
```

```python
# Fill missing values with "Unknown" in string/object columns
# where the missing percentage is less than or equal to 50%
df["first_review"] = df["first_review"].fillna("Unknown")
df["last_review"] = df["last_review"].fillna("Unknown")
```

```python
# Select all numeric columns (integer and float)
numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns

# Fill missing values in numeric columns with the median value
for col in numeric_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())
```

```python
# --- Calculate Missing Percentage ---
print("\n  MISSING VALUES PER COLUMN:")
missing_percent = df.isnull().mean() * 100
missing_cols = missing_percent[missing_percent > 0].sort_values(ascending=False)

if len(missing_cols) > 0:
    print(missing_cols)
    print(f"\nTotal columns with missing data: {len(missing_cols)}")
else:
    print("No missing values found in any column!")
```

```python
# checking datatype of all columns
# checking top 5 rows of all columns
print(df["price"].dtype)
print(df["price"].head())
```

```python
# Convert price from string/object to numeric format
# Remove "$" and "," symbols before converting
df["price"] = (
    df["price"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)

# Fill missing price values with the median price
df["price"] = df["price"].fillna(df["price"].median())
```
